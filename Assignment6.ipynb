## Q.1
Consider the following sensor readings:

`temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])`

A sensor has an average measurement error of +2°C. Perform the following **without using a for loop**:

a. Add 2°C to every reading using vectorization.  
b. Convert all temperatures from Celsius to Fahrenheit using F = (9/5)C + 32.  
c. Identify all readings greater than 32°C using Boolean indexing.  
d. Count how many readings exceed 32°C.  
e. Explain how vectorization and Boolean indexing can make data processing more efficient than explicitly iterating through every value.

````python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Original readings:", temperature)
````

````python
# a) Add 2 °C to every reading (vectorized)
adjusted = temperature + 2
print("a) After adding 2°C        :", adjusted)
````

````python
# b) Celsius -> Fahrenheit
fahrenheit = (9/5) * temperature + 32
print("b) Temperatures in °F      :", fahrenheit)
````

````python
# c) Boolean indexing: readings > 32°C
mask = temperature > 32
print("Boolean mask               :", mask)
print("c) Readings greater than 32:", temperature[mask])
````

````python
# d) Count of readings above 32°C
count = np.sum(temperature > 32)
print("d) Number of readings > 32°C:", count)
````

**e) Explanation:**

- **Vectorization** applies an operation to the whole array at once. The loop runs inside NumPy's optimized C code instead of Python's interpreter, so there is no per-element Python overhead. Arrays are stored in contiguous memory with a single data type, which also makes the operations cache-friendly and able to use CPU SIMD instructions. This is typically tens to hundreds of times faster than a Python `for` loop.
- **Boolean indexing** replaces an `if` inside a loop with a single expression such as `temperature > 32`, which builds a True/False mask for all elements together. The mask can then select values (`temperature[mask]`) or be counted (`np.sum(mask)`) in one step.
- The code is also shorter, easier to read and less error-prone (no manual indices, no `append`), and it scales well to millions of readings.

## Q.2
The following matrix represents the number of daily steps recorded by four users over three days:

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]])
```

a. Calculate the total steps recorded.  
b. Calculate the mean number of steps.  
c. Find the maximum and minimum values.  
d. Calculate the total steps for each day using `axis=0`.  
e. Calculate the total steps for each user using `axis=1`.  
f. Find the user/day position containing the maximum number of steps using `argmax()`.

````python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]])
print(steps)
````

````python
print("a) Total steps          :", np.sum(steps))
print("b) Mean steps           :", np.mean(steps))
print("c) Maximum              :", np.max(steps))
print("   Minimum              :", np.min(steps))
````

````python
# d) Total steps for each day (sum down the rows -> one value per column)
print("d) Total per day  (axis=0):", np.sum(steps, axis=0))

# e) Total steps for each user (sum across the columns -> one value per row)
print("e) Total per user (axis=1):", np.sum(steps, axis=1))
````

````python
# f) Position of the maximum value
flat_index = np.argmax(steps)
user, day = np.unravel_index(flat_index, steps.shape)
print("f) Flat index from argmax():", flat_index)
print(f"   Position (row, col)     : ({user}, {day})")
print(f"   -> User {user + 1}, Day {day + 1} with {steps[user, day]} steps")
````

## Q.3
Write a Python program using NumPy to perform the following operations:

a. Create the array `original = np.array([1, 2, 3, 4, 5, 6])`.  
b. Create a slice containing elements from index 1 to 4 and store it in `subset`.  
c. Modify the first element of `subset` to 999 and display both `original` and `subset`.  
d. Create another slice from `original` using `.copy()`. Modify its first element to 500 and display `original` and the copied array.  
e. Create an array of numbers 1 to 12 using `np.arange()` and reshape it into a 3 × 4 matrix.  
f. From the 3 × 4 matrix, extract: first row, last row, second column, and elements from rows 1–2 and columns 2–3.  
g. Create a flattened version of the matrix using both `flatten()` and `ravel()`.  
h. Modify an element in the array returned by `ravel()` and observe whether the original is affected.  
i. Modify an element in the array returned by `flatten()` and observe whether the original is affected.  
j. Display the shape, ndim, size and dtype of the original 3 × 4 matrix.

````python
# a) Create the array
original = np.array([1, 2, 3, 4, 5, 6])
print("original:", original)
````

````python
# b) Slice from index 1 to 4 (indices 1, 2, 3 -> end index is exclusive)
subset = original[1:4]
print("subset  :", subset)
````

````python
# c) Modify subset[0] -> slice is a VIEW, so original changes too
subset[0] = 999
print("original:", original)
print("subset  :", subset)
````

A basic slice is a **view** of the original data (no new memory), so changing `subset` also changed `original`.

````python
# d) Slice with .copy() -> independent array
copied = original[1:4].copy()
copied[0] = 500
print("original:", original)
print("copied  :", copied)
````

`.copy()` creates a separate array, so changing it does **not** affect `original`.

````python
# e) arange + reshape
matrix = np.arange(1, 13).reshape(3, 4)
print(matrix)
````

````python
# f) Indexing and slicing
print("First row            :", matrix[0])
print("Last row             :", matrix[-1])
print("Second column        :", matrix[:, 1])
print("Rows 1-2, cols 2-3   :\n", matrix[0:2, 1:3])
````

*Note: rows 1–2 and columns 2–3 are taken as the 1st–2nd rows and 2nd–3rd columns (1-based), i.e. `matrix[0:2, 1:3]`.*

````python
# g) flatten() and ravel()
flat = matrix.flatten()
rav = matrix.ravel()
print("flatten():", flat)
print("ravel()  :", rav)
````

````python
# h) Modify ravel() result -> original affected?
rav[0] = 100
print("ravel array :", rav)
print("matrix after modifying ravel():\n", matrix)
````

`ravel()` returns a **view** whenever possible, so the original matrix **was affected** (element `[0, 0]` became 100).

````python
# i) Modify flatten() result -> original affected?
flat[1] = 200
print("flatten array:", flat)
print("matrix after modifying flatten():\n", matrix)
````

`flatten()` always returns a **copy**, so the original matrix was **not affected** by changing `flat[1]`.

````python
# j) Attributes of the 3 x 4 matrix
print("Shape:", matrix.shape)
print("ndim :", matrix.ndim)
print("size :", matrix.size)
print("dtype:", matrix.dtype)
````

## Q.4
A simple cognitive-assistive system wants to predict a user's assistance score based on three factors: **sleep hours**, **activity level** and **stress level**.

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])
```

Each row is one user and each column is a feature.

a. Display the shape and dimensions of X.  
b. Calculate `X.T` and explain what the transpose represents.  
c. Calculate the matrix product `X.T @ X`.  
d. Calculate the matrix inverse using `np.linalg.inv()` where appropriate.  
e. Implement the Ordinary Least Squares equation β = (XᵀX)⁻¹Xᵀy using NumPy.  
f. Explain what the resulting coefficients represent in the context of the three user features.  
g. For `new_user = np.array([5, 40, 7])`, use the calculated coefficients to predict the assistance score.

````python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

# a) Shape and dimensions
print("Shape of X     :", X.shape)
print("Dimensions (ndim):", X.ndim)
````

````python
# b) Transpose
print("X.T =\n", X.T)
print("Shape of X.T:", X.T.shape)
````

**Explanation (b):** The transpose swaps rows and columns. `X` has shape (4 × 3): 4 users × 3 features. `X.T` has shape (3 × 4): each **row** now holds one **feature** (sleep, activity, stress) across all four users, and each column is one user. This arrangement is what lets `X.T @ X` combine the features with each other.

````python
# c) Matrix product X.T @ X  (3x4 @ 4x3 -> 3x3)
XtX = X.T @ X
print("X.T @ X =\n", XtX)
print("Shape:", XtX.shape)
````

````python
# d) Inverse of X.T @ X  (a square matrix, so the inverse is valid if det != 0)
print("Determinant of X.T @ X:", np.linalg.det(XtX))
XtX_inv = np.linalg.inv(XtX)
print("(X.T @ X)^-1 =\n", XtX_inv)

# Check: matrix times its inverse should give the identity matrix
print("\nCheck (X.T @ X) @ inverse ≈ I:\n", np.round(XtX @ XtX_inv, 6))
````

*Note:* `np.linalg.inv()` is applied to `X.T @ X` (3 × 3), **not** to `X` itself. `X` is 4 × 3 (not square), so it has no inverse. The determinant is non-zero, so `X.T @ X` is invertible.

````python
# e) Ordinary Least Squares: beta = (X^T X)^-1 X^T y
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("OLS coefficients (beta):", beta)
print()
print("Sleep hours    :", round(beta[0], 4))
print("Activity level :", round(beta[1], 4))
print("Stress level   :", round(beta[2], 4))
````

**Explanation (f):** Each coefficient is the change in the predicted assistance score for a **one-unit increase** in that feature, holding the other two features fixed.

- **Sleep hours (≈ −0.16):** a very small negative weight, so extra sleep slightly lowers the predicted assistance need.
- **Activity level (≈ +0.14):** a small positive weight. This is counter-intuitive, but activity is measured on a much larger scale (30–80) than the others, and with only 4 data points features that move together (high sleep, high activity, low stress) can get odd signs.
- **Stress level (≈ +10.06):** by far the largest weight, so stress is the dominant predictor, and each extra stress point raises the predicted score by about 10.

The model has **no intercept term** (no bias column), so it is forced through the origin, and with only 4 samples these coefficients are illustrative rather than reliable.

````python
# g) Predict for a new user
new_user = np.array([5, 40, 7])
prediction = new_user @ beta
print("New user features :", new_user)
print("Predicted assistance score:", round(prediction, 2))
````

**Result:** The predicted assistance score for the new user is about **75.28**. This is high, driven mainly by the user's high stress level (7).
