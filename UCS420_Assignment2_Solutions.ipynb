## Q1. Lists

```python
# Enter your roll number
roll_no = ("1024170019")

# Extract individual digits and multiply each by 10
digits = [int(d) for d in roll_no]
L = [d * 10 for d in digits]

print("Initial L:", L)

# ii. Add two numbers: one with append() and one with insert()
L.append(100)
print("After append(100):", L)  # append() adds the value at the end

L.insert(1, 50)
print("After insert(1, 50):", L)  # insert() places 50 at index 1

# iii. Remove two elements
L.remove(50)
print("After remove(50):", L)  # remove() removes the first matching value

removed = L.pop()
print("After pop():", L)
print("Element removed by pop():", removed)  # pop() removes and returns an element

# iv. Sort ascending and descending
L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)

# v. Slicing
print("First three:", L[:3])
print("Last three:", L[-3:])

# vi. Elements greater than the average
average = sum(L) / len(L)
greater_than_average = [x for x in L if x > average]

print("Average:", average)
print("Elements greater than average:", greater_than_average)
```

## Q2. Tuples



```python
# Use the first 8 values from L
scores = tuple(L[:8])

# i. Highest score and its index
highest = max(scores)
highest_index = scores.index(highest)

# Lowest score and how many times it appears
lowest = min(scores)
lowest_count = scores.count(lowest)

print("scores:", scores)
print("Highest score:", highest)
print("Index of highest:", highest_index)
print("Lowest score:", lowest)
print("Lowest appears:", lowest_count, "time(s)")

# ii. Reverse tuple and return as a list
# Tuples are immutable, so they cannot be reversed in place.
reversed_scores = list(reversed(scores))
print("Reversed tuple as list:", reversed_scores)

# iii. Search for a score
value = int(input("Enter a score to search: "))

if value in scores:
    print("First occurrence index:", scores.index(value))
else:
    print("Score is not present in the tuple.")

# iv. Attempt to modify tuple
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)
    print("A tuple is immutable, so its elements cannot be changed directly.")

# v. Tuple unpacking using *
if len(scores) >= 2:
    first, second, *remaining = scores
    print("First:", first)
    print("Second:", second)
    print("Remaining:", remaining)
```

## Q3. Random Numbers



```python
import random
from collections import Counter

# Use your roll number as the seed
seed = int(input("Enter your roll number for random.seed(): "))
random.seed(seed)

# i. Generate 100 random numbers between 100 and 900 inclusive
numbers = [random.randint(100, 900) for _ in range(100)]

print("Random numbers:")
print(numbers)

# ii. Count and print odd numbers
odd_numbers = [x for x in numbers if x % 2 != 0]
print("\nNumber of odd numbers:", len(odd_numbers))
print("Odd numbers:", odd_numbers)

# iii. Count and print even numbers
even_numbers = [x for x in numbers if x % 2 == 0]
print("\nNumber of even numbers:", len(even_numbers))
print("Even numbers:", even_numbers)

# Helper function to check prime
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True

# iv. Count and list prime numbers
prime_numbers = [x for x in numbers if is_prime(x)]

print("\nNumber of prime numbers:", len(prime_numbers))
print("Prime numbers:", prime_numbers)

# v. Most frequent number and its frequency
frequency = Counter(numbers)
most_frequent_number, most_frequent_count = frequency.most_common(1)[0]

print("\nMost frequent number:", most_frequent_number)
print("Frequency:", most_frequent_count)
```

## Q4. Sets

Using the same digits of the roll number, create:
- `A = {digit × 7}`
- `B = {digit × 9}`

Then perform union, intersection, difference, symmetric difference, subset/superset checks, and `discard()`.

```python
# Use the original digits from Q1
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("A =", A)
print("B =", B)

# vi. Union
print("\nA union B =", A.union(B))

# vii. Intersection
print("A intersection B =", A.intersection(B))

# viii. Difference
print("A - B =", A.difference(B))
print("B - A =", B.difference(A))
print("difference() gives directional differences; symmetric_difference() gives values in either set but not both.")

# ix. Symmetric difference
print("Symmetric difference =", A.symmetric_difference(B))

# x. Subset and superset
print("\nIs A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

# xi. discard()
X = int(input("\nEnter X to discard from A: "))
A.discard(X)

print("A after discard(X):", A)
print("discard() is safer than remove() because it does not raise an error if X is absent.")
```

## Q5. Dictionaries


```python
# Enter your details
my_dict = {
    "name": input("Enter your name: "),
    "roll_no": input("Enter your roll number: "),
    "branch": input("Enter your branch: "),
    "age": int(input("Enter your age: ")),
    "city": input("Enter your home city: ")
}

print("Original dictionary:")
print(my_dict)

# i. Rename city -> location without recreating the whole dictionary
my_dict["location"] = my_dict.pop("city")
print("\nAfter renaming city to location:")
print(my_dict)

# ii. Add CGPA
my_dict["cgpa"] = float(input("Enter your CGPA: "))
print("\nAfter adding CGPA:")
print(my_dict)

# iii. Increase age by 1
my_dict["age"] += 1
print("\nAfter increasing age by 1:")
print(my_dict)

# iv. Delete branch using pop() and del in separate copies
dict_pop = my_dict.copy()
removed_branch = dict_pop.pop("branch")
print("\nUsing pop():")
print(dict_pop)
print("pop() returns the removed value:", removed_branch)

dict_del = my_dict.copy()
del dict_del["branch"]
print("\nUsing del:")
print(dict_del)
print("del removes the key but does not return its value.")

# v. Iterate using items()
print("\nKey-value pairs:")
for key, value in my_dict.items():
    print(f"{key} → {value}")

# vi. Safely check for email
if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail key does not exist.")

# vii. Merge with a fictional friend's dictionary
friend_dict = {
    "name": "Friend",
    "roll_no": "999999",
    "branch": "CSE",
    "age": 20,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print("\nMerged dictionary:")
print(merged_dict)

print("When both dictionaries have the same key, the value from the second dictionary wins.")

# viii. Dictionary comprehension for string-valued pairs
string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nOnly key-value pairs whose values are strings:")
print(string_values)
```
