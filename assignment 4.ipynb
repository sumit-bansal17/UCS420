**b)** Find the most frequent value and their index(es) in the following arrays:

i. `x = np.array([1,2,3,4,5,1,2,1,1,1])`  
ii. `y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])`

```python
def most_frequent(a):
    values, counts = np.unique(a, return_counts=True)
    val = values[np.argmax(counts)]
    idx = np.where(a == val)[0]
    return val, idx

x = np.array([1,2,3,4,5,1,2,1,1,1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

for name, a in (("x", x), ("y", y)):
    v, i = most_frequent(a)
    print(f"Array {name} = {a}")
    print(f"  Most frequent value : {v}")
    print(f"  Index(es)           : {i}\n")

# Note: in y, the values 1 and 3 both occur 3 times (a tie); np.argmax picks the first one (1).
vals, cnts = np.unique(y, return_counts=True)
print("Tie check for y ->", dict(zip(vals, cnts)))
print("All values with max count in y:", vals[cnts == cnts.max()])
for v in vals[cnts == cnts.max()]:
    print(f"  Indices of {v}:", np.where(y == v)[0])
```

## Q.4
Write a program to create a 1-D NumPy array named `sumit_bansal` with evenly spaced 25 numbers from 10 to 100 using `linspace()`. Print the dimensions, shape, total elements, data type of each element and total number of bytes consumed by the array. Find the transpose of this array using `reshape()`. Can we do the same with the `T` attribute?

```python
sumit_bansal = np.linspace(10, 100, 25)

print("Array            :", sumit_bansal)
print("Dimensions (ndim):", sumit_bansal.ndim)
print("Shape            :", sumit_bansal.shape)
print("Total elements   :", sumit_bansal.size)
print("Data type        :", sumit_bansal.dtype)
print("Total bytes      :", sumit_bansal.nbytes)
```

```python
# Transpose using reshape(): 1-D (25,) -> column vector (25, 1)
transposed = sumit_bansal.reshape(25, 1)
print("Shape after reshape(25, 1):", transposed.shape)
print(transposed)
```

```python
# Using the T attribute on the 1-D array
print("Shape of sumit_bansal.T :", sumit_bansal.T.shape)
print("Same as original?    :", np.array_equal(sumit_bansal.T, sumit_bansal))

# T works only if the array is first made 2-D:
print("Shape of sumit_bansal.reshape(1, 25).T :", sumit_bansal.reshape(1, 25).T.shape)
```

**Answer:** No. A 1-D array has only one axis, so the `T` attribute returns the same array unchanged (shape stays `(25,)`). To get a real transpose (column vector), the array must first be 2-D, e.g. `reshape(1, 25).T`, or use `reshape(25, 1)` directly.

## Q.5
Create a 2-D array of three rows and four columns, named `ucs420_<sumit_bansal>` with the values 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it `reshaped_ucs420_<sumit_bansal>`. Resize the array to two rows and three columns and name it `resized_ucs420_<sumit_bansal>`.

```python
ucs420_sumit_bansal = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]).reshape(3, 4)
print("ucs420_sumit_bansal:\n", ucs420_sumit_bansal)
print()
print("Mean            :", np.mean(ucs420_sumit_bansal))
print("Median          :", np.median(ucs420_sumit_bansal))
print("Max             :", np.max(ucs420_sumit_bansal))
print("Min             :", np.min(ucs420_sumit_bansal))
print("Unique elements :", np.unique(ucs420_sumit_bansal))
```

```python
reshaped_ucs420_sumit_bansal = ucs420_sumit_bansal.reshape(4, 3)
print("reshaped_ucs420_sumit_bansal (4x3):\n", reshaped_ucs420_sumit_bansal)
```

```python
resized_ucs420_sumit_bansal = np.resize(ucs420_sumit_bansal, (2, 3))
print("resized_ucs420_sumit_bansal (2x3):\n", resized_ucs420_sumit_bansal)
```
